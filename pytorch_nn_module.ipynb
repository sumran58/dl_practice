```python
import torch
import torch.nn as nn

class Model(nn.Module):
  def __init__(self,num_features):
    super().__init__() #Model is a sub class becuase it is calling the fetuares of the nn.Module so with th help of super keywrod we are calling the init function of the Module class which is the super class
    self.linear=nn.Linear(num_features,1) #bracket me inpout me num_fetaures and output me 1 output
    self.sigmoid=nn.Sigmoid()

  def forward(self,features):
    out=self.linear(features)
    out=self.sigmoid(out)
    return out
```

```python
features=torch.rand(10,5)
model=Model( features.shape[1] )
model(features)
```

```python
import torch
import torch.nn as nn

class Model(nn.Module):
  def __init__(self,num_features):
    super().__init__() #Model is a sub class becuase it is calling the fetuares of the nn.Module so with th help of super keywrod we are calling the init function of the Module class which is the super class
    self.linear1=nn.Linear(num_features,3) #bracket me inpout me num_fetaures and output me 1 output
    self.relu=nn.ReLU()
    self.linear2=nn.Linear(3,1)
    self.sigmoid=nn.Sigmoid()

  def forward(self,features):
    out=self.linear1(features)
    out=self.relu(out)
    out=self.linear2(out)
    out=self.sigmoid
    return out
```

```python
features=torch.rand(10,5)
model=Model( features.shape[1] )
model(features)
```
