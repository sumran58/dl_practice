##inplace operation

```python
#koi bhi 2 tensor ka kuch bhi operation kiye to wo new tensor me store hota that means it takes the memory and we dont want that because in some places it might
#take huge amount of memory so for that we have the function we underscore
m=torch.randint(size=(2,2),low=0,high=10)
n=torch.randint(size=(2,2),low=0,high=10)
```

```python
m.add_(n) #this wil perform the addition operation but it will save the chnages in the m varaible at that particular memory and will not occupy the different memory
```

```python
m.relu_() #uderscore represent the inplace
```
